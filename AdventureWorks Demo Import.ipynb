## AdventureWorks Demo Import
Import the CSV-files from the AdventureWorks demo database as Delta tables in Databricks.
This notebook assumes you already placed the AdventureWorks demo CSV files to DBFS.



```python
import os

# Path to the CSV-files on DBFS
base_path = "/Volumes/workspace/default/adventureworks/"

# Lijst van tabellen om te importeren
tables = [
    "dim_customer",
    "dim_product",
    "dim_employee",
    "dim_sales_territory",
    "dim_date",
    "fact_internet_sales",
    "fact_reseller_sales",
    "dim_reseller",
    "dim_promotion",
    "dim_currency",
]

for table in tables:
    path = f"{base_path}/{table}.csv"
    df = spark.read.option("header", True).csv(path)
    df.write.mode("overwrite").saveAsTable(f"adventureworks.{table}")
    print(f"✅ Created table: adventureworks.{table}")
```
